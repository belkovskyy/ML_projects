#### Избавление от лишних столбцов и оформление выборок

В первую очередь избавимся от ненужных столбцов - NumberOfPictures и PostalCode.

Придется также избавиться и от столбцов даты - DateCrawled, DateCreated и LastSeen (Но при этом сохраним RegistrationYear).

```python
data = data.drop(['NumberOfPictures', 'PostalCode', 'DateCrawled', 'DateCreated', 'LastSeen'], axis=1)
data
```

Выделим признаки. Поделим датасет на тренировочный (75%) и тестовый (25%). Валидацию будем проводить с помощью кросс-валидации.

```python
features = data.drop(['Price'], axis=1)
target = data['Price']

#features_train, features_valid, target_train, target_valid = train_test_split(
    #features, target, test_size=0.4, random_state=12345)
#features_valid, features_test, target_valid, target_test = train_test_split(
    #features_valid, target_valid, test_size=0.5, random_state=12345)
    
features_train, features_test, target_train, target_test = train_test_split(
     features, target, test_size=0.2, random_state=12345)
```

Проверим.

```python
print(features_train.shape)
print(target_train.shape)

#print(features_valid.shape)
#print(target_valid.shape)

print(features_test.shape)
print(target_test.shape)
```

#### Применение кодирования

Будем комбинировать два типа кодирования - порядковое (Ordinal) и прямое (OHE).

Применим кодирование типа OHE к категориальным признакам, не имеющим множество уникальных значений (ко всем категориальным признакам кроме Brand и Model), чтобы не создавать много столбцов.

```python
ohe_features_ridge = features_train.select_dtypes(include='object').columns.to_list()
ohe_features_ridge.remove('Brand')
ohe_features_ridge.remove('Model')
ohe_features_ridge
```

```python
features_train_ridge = features_train.copy()
#features_valid_ridge = features_valid.copy()
features_test_ridge = features_test.copy()
```

```python
encoder_ohe = OneHotEncoder(drop='first', sparse=False)
encoder_ohe.fit(features_train_ridge[ohe_features_ridge])
```

```python
features_train[
    encoder_ohe.get_feature_names()
] = encoder_ohe.transform(features_train_ridge[ohe_features_ridge])

features_train = features_train.drop(ohe_features_ridge, axis=1)

features_train
```

```python
# features_valid[
#     encoder_ohe.get_feature_names()
# ] = encoder_ohe.transform(features_valid_ridge[ohe_features_ridge])

# features_valid = features_valid.drop(ohe_features_ridge, axis=1)

# features_valid
```

```python
features_test[
    encoder_ohe.get_feature_names()
] = encoder_ohe.transform(features_test_ridge[ohe_features_ridge])

features_test = features_test.drop(ohe_features_ridge, axis=1)

features_test
```

Количество столбцов увеличилось незначительно.

В свою очередь, к признакам с множеством значений применим OrdinalEncoder. Если не планируется использование модели линейной регрессии градация в столбцах не должна навредить предсказаниям.

```python
encoder_lab = LabelEncoder()
features_train['Model'] = encoder_lab.fit_transform(features_train['Model'])
features_train['Brand'] = encoder_lab.fit_transform(features_train['Brand'])

# features_valid['Model'] = encoder_lab.fit_transform(features_valid['Model'])
# features_valid['Brand'] = encoder_lab.fit_transform(features_valid['Brand'])

features_test['Model'] = encoder_lab.fit_transform(features_test['Model'])
features_test['Brand'] = encoder_lab.fit_transform(features_test['Brand'])


features_train
```

Помимо этого признаки с порядковым кодированием и оставшиеся количественные признаки замасштабируем.

```python
numeric = ['RegistrationYear', 'Power', 'Kilometer', 'RegistrationMonth', 'Model', 'Brand']

scaler = StandardScaler()
features_train[numeric] = scaler.fit_transform(features_train[numeric])
#scaler.fit(features_train[numeric])
#features_train[numeric] = scaler.transform(features_train[numeric])
#features_valid[numeric] = scaler.transform(features_valid[numeric])
features_test[numeric] = scaler.transform(features_test[numeric])
pd.options.mode.chained_assignment = None
```

### Обучение

Займемся обучением моделей. Рассмотрим 2 модели градиентного бустинга - LGBM и Catboost, а также модель случайного леса.

Для леса и LGBM будем использовать GridSearchCV, а для кэтбуста встроенные функции.

```python
score_rmse = make_scorer(mean_squared_error, greater_is_better=False, squared=False)
```

#### LGBM

Начнем с модели LGBM. Зададим варианты гиперпараметров и обозначим саму модель.

```python
parameters_lgbm = {'learning_rate': [0.15, 0.2],
    'max_depth': [6, 8],
    'n_estimators': [150, 200]}
    #num_leaves
model_lgbm = LGBMRegressor(random_state=12345) # objective='regression',
```

Подберем наилучшие гиперпараметры методом GridSearch, проведем кросс-валидацию и оценим время.

```python
# lgbm_cross_val = cross_val_score(estimator=model_lgbm, X=features_train, y=target_train, cv=3, scoring=score_rmse)
# lgbm_cross_val.mean()
```

```python
%%time

gs_lgbm = GridSearchCV(estimator=model_lgbm, param_grid=parameters_lgbm, cv=3, n_jobs=-1, scoring=score_rmse).fit(features_train, target_train) #
```

Оценим лучшие гиперпараметры и лучший результат.

```python
gs_lgbm.best_params_
```

```python
abs(gs_lgbm.best_score_)
```

Сохраним лучший вариант.

```python
best_model_lgbm = gs_lgbm.best_estimator_
```

#### CatBoost

С кэтбустом поступим немного по-другому. В кэтбусте есть свой grid_search и кросс-валидация, мы сначала найдем наилучшие гиперпараметры с помощью grid_search, а потом проведем тестирование потенциала с помощью кросс-валидации.

```python
parameters_cat = {
        #'iterations': [100, 150, 200],
        'learning_rate': [0.15, 0.2],
        'depth': [8, 10],
        #'l2_leaf_reg': [1, 3, 5, 7, 9],
        
}
                  
cat = CatBoostRegressor(loss_function = 'RMSE', random_state = 12345)
```

```python
%%time

gs_cat = cat.grid_search(parameters_cat, features_train, target_train, plot=True)
```

Стоит также отметить показатели графика в grid_search'е от кэтбуст. Предусматриваются обучающий и тестовый график, и, хотя показатели RMSE по по обучающему графику стабильно значимо падают, нельзя то же самое сказать про тестовый. Таким образом можно наблюдать за возможным переобучением модели. Например, можем видеть, что learning rate равный 0.2 выдает не отличный от 0.15 результат.

```python
gs_cat['params']
```

Также проведем кросс-валидацию по найденным гиперпараметрам.

```python
train_dataset = Pool(features_train, target_train) 
#test_dataset = Pool(features_test, target_test)
```

```python
params_cat = {
          "depth": 10,
          "eval_metric": "RMSE",
          "loss_function": "RMSE",
          "learning_rate": 0.15,
          "l2_leaf_reg": 3,
          "verbose": 100}

# cat_cross = CatBoostRegressor()

# cv_dataset = Pool(data=features_train,
#                   label=target_train)

scores = cv(params = params_cat,
            pool = train_dataset,
            fold_count=3,
            shuffle = True,
            plot="True",
            partition_random_seed = 0)
```

Таким образом, валидационный результат RMSE на лучших параметрах - 1453. 

```python
scores['test-RMSE-mean']
```

```python
scores[scores['test-RMSE-mean'] == scores['test-RMSE-mean'].min()]
```

Переобучения не случилось, лучшие результаты наблюдаются на последних итерациях.

Безоговорочным лидером является CatBoost, он уступает по времени обучения LGBM, но выдает гораздо более качественный результат RMSE.

На втором месте - LGBM. Но стоит отметить, что у этой модели наибольшая разница между RMSE при валидации и финальным RMSE - где-то на 60 пунктов. Возможно произошло переобучение и стоило бы поработать с гиперпараметрами. Жаль, что в GridSearchCV не предусматриваются отдельные графики для обучения и теста, чтобы это фиксировать. 

Модель случайного леса, которая, как правило, выдавала более лучшие результаты, чем модели линейной регрессии и дерева решений, значимо уступает моделям градиентного бустинга, особенно в плане времени обучения. Тем не менее, стоит отметить, что разница у RMSE при валидации и финального RMSE минимальна, возможно модель можно было бы еще дообучать, чтобы добиться результатов получше, но тогда бы это заняло еще больше времени.

Гигантский результат дамми-модели доказывает качество полученных нами моделей и пользу машинного обучения в данном вопросе.

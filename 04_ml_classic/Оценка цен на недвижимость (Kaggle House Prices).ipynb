Рассмотрим 4 модели. Три модели градиентого бустинга (так как градиентный бустинг, зачастую, наиболе эффективно справляется с табличными данными) - CatBoost, LightGBM и XGBoost, а также отдельно рассмотрим модель линейной регрессии, которая, несмотря на свою простоту, вполне может использоваться в вопросе прогнозирования цен.
